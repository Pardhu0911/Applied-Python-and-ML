```python
# Do NOT use import more packages than what are given 
import json
import string
import matplotlib.pyplot as plt


# Exercise 1: Word Frequency Analysis in "Pride and Prejudice"
'''
Read the "Pride and Prejudice" text file.
Clean the text by removing punctuations using string.punctuation
Clean the text by removing stop words from the provided stop words list in "stopwords.txt".
Count the frequency of each word in the cleaned text.
Save the word frequency data as a JSON file.
'''

##############################################################
##############  Complete your code below #####################
# Step 1: Read the text file and clean the data
with open('pride_and_prejudice.txt', 'r', encoding='utf-8') as f:
    text = f.read().lower()

# Step 2: Remove punctuation
for char in string.punctuation:
    text = text.replace(char, ' ')
words = text.split()

# Step 3: Remove stop words
with open('stopwords.txt', 'r') as f:
    stop_words = f.read().split()

# Step 4: Store the cleaned text in a list
cleaned_text = [word for word in words if word not in stop_words and word.isalpha()]

# Step 5: Count the frequency of each word
word_freq = {}
for word in cleaned_text:
    word_freq[word] = word_freq.get(word, 0) + 1

# Step 6: Save the word frequency dictionary to 'word_frenquency.json' as a JSON file
with open('word_frenquency.json', 'w') as f:
    json.dump(word_freq, f, indent=4)
##############################################################


# Test the content of the dictionary
# (Note: word_freq must be accessible here for testing)
for _ in range(3):
    print(list(word_freq.items())[-1 - _])

# Expected output
'''
('uniting', 1)
('city', 1)
('pollution', 1)
'''
```

```python
# Exercise 2: Statistical Analysis and Visualization
'''   
Perform a statistical analysis of word lengths in the Pride and Prejudice text. 
Calculate the average word length, the most common word length, 
and the distribution of word lengths. 
Visualize the distribution using a simple bar chart.
'''

##########################################################
#####     Complete your code below                ########
#         Use the cleaned text from Exercise 1
#         Calculate the statistics including
#         1. average length of words 
total_length = sum(len(word) for word in cleaned_text)
average_length = total_length / len(cleaned_text)

#         2. most common word length in the book
length_distribution = {}
for word in cleaned_text:
    length = len(word)
    length_distribution[length] = length_distribution.get(length, 0) + 1

#         3. store the length distribution in length_distribution dictionary
max_key = max(length_distribution, key=length_distribution.get)
###########################################################


# Visualize the distribution of word lengths
plt.bar(length_distribution.keys(), length_distribution.values())
plt.xlabel('Word Length')
plt.ylabel('Frequency')
plt.title('Distribution of Word Lengths in Pride and Prejudice')
plt.show()


# Test the result
print(f"Average Word Length: {average_length}")
print(f"Most Common Word Length: {max_key}")
for _ in range(3):
    print(list(length_distribution.items())[-1 - _])

# Expected output
'''   
Average Word Length: 6.754793315743184
Most Common Word Length: 6
(26, 1)
(23, 1)
(18, 3)
'''
```

```python
# Exercise 3: Character Name Frequency Analysis and Visualization

# Define a list of main character names
characters = ['elizabeth', 'darcy', 'bennet', 'jane', 'bingley', 'lydia', 'wickham', 'collins']

##################################################
######  Complete your code below   ###############
# Step 1: Count the frequency of each character name
character_freq = {name: word_freq.get(name, 0) for name in characters}

# Step 2: Save the results as a JSON file
with open('character_frequency.json', 'w') as f:
    json.dump(character_freq, f, indent=4)
##################################################


# Visualize the results using a pie chart
plt.pie(character_freq.values(), labels=character_freq.keys(), autopct='%1.1f%%')
plt.title('Character Name Frequency in Pride and Prejudice')
plt.show()
```
