```python
# Exercise 2: Dictionary Operations
# Create random_dict that contains 100 random letter-number pairs
random.seed(26)
alphabet = 'abcdefghijklmnopqrstuvwxyz'
random_dict = {letter: random.randint(1, 100) for letter in alphabet}

s = "compare python basic data types"
# Calculate the sum of the values corresponding to each letter in the s, using random_dict

sum(random_dict.get(ch,0) for ch in s if ch.isalpha()) # complete the code within the parentheses 
```
