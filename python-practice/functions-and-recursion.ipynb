```python
# Module 4 Assignment: Advanced Functions

## Exercise 1: Complex Function Design and Scope Management

# Function to analyze student data and calculate averages
def data_analyzer(students_scores):
    """
    Analyzes student scores and returns averages, flagging those below the threshold.
    """
    result = {}
    for name, scores in students_scores:
        # Calculate the average of the list of scores
        avg = sum(scores) / len(scores)
        
        # Check against the global variable 'threshold'
        if avg < threshold:
            # Include warning message if below threshold
            result[name] = (avg, 'Below Threshold')
        else:
            # Return average in a tuple to match provided output format
            result[name] = (avg,)
            
    return result

# Example usage for Exercise 1 using a threshold of 70
threshold = 70  # Global variable for threshold
students_scores = [("Alice", [85, 92, 78]), ("Bob", [90, 88, 95]), ("Charlie", [60, 58, 55])]
data_analyzer(students_scores)
```

```python
## Exercise 2: Recursion 

# Function to return all permutations of a string using recursion
import itertools

def my_permute(s):
    """
    Recursively generates all permutations of a string s.
    Returns a list of tuples to match itertools format.
    """
    # Base case: if the string is empty or one character
    if len(s) <= 1:
        return [tuple(s)]
    
    perms = []
    for i in range(len(s)):
        char = s[i]
        # Get the string remaining after removing the current character
        remaining = s[:i] + s[i+1:]
        
        # Recursively find permutations of the remaining string
        for p in my_permute(remaining):
            perms.append((char,) + p)
            
    return perms

# Compare your solution and using itertools.permutations()
test_string = "Test12"
# This comparison will now return True
list(itertools.permutations(test_string)) == my_permute(test_string)
```
