```python
# Module 5 Classes and Objects (OOP)

# Exercise 1: Creating Classes and Objects

# Define a class 'Vehicle' to represent a generic vehicle
class Vehicle:
    
    # Initialize the attributes for make, model, year, and mileage
    def __init__(self, make, model, year, mileage):
        self.make = make
        self.model = model
        self.year = year
        self.mileage = mileage
    
    # Method to display vehicle information
    def display_info(self):
        print(f"Make: {self.make}")
        print(f"Model: {self.model}")
        print(f"Year: {self.year}")
        print(f"Mileage: {self.mileage} miles")

# Create an instance of Vehicle
car = Vehicle("Honda", "Civic", 2020, 15000)

# Call the display_info method to print details of the vehicle
car.display_info()
```

```python
# Exercise 2: Inheritance and Method Overriding

# Define a class 'ElectricVehicle' that inherits from the 'Vehicle' class
class ElectricVehicle(Vehicle):
    def __init__(self, make, model, year, mileage, battery_size, range_per_charge):
        # Call the parent class constructor to initialize common attributes
        super().__init__(make, model, year, mileage)
        # Initialize additional attributes specific to electric vehicles
        self.battery_size = battery_size
        self.range_per_charge = range_per_charge

    # Override the display_info method to include battery and range details
    def display_info(self):
        # Call the parent class method to display common details
        super().display_info()
        # Display additional electric vehicle-specific details
        print(f"Battery Size: {self.battery_size} kWh")
        print(f"Range per Charge: {self.range_per_charge} miles")

# Create an instance of ElectricVehicle
tesla = ElectricVehicle("Tesla", "Model S", 2022, 10000, 100, 370)

# Call the display_info method to print details of the electric vehicle
tesla.display_info()
```

```python
# Exercise 3: Polymorphism

# Define a show_vehicle_details() function that 
# accepts any vehicle object and displays its details
def show_vehicle_details(vehicle_obj):
    vehicle_obj.display_info()

# Demonstrating polymorphism by passing different types of vehicles to the same function
print("\nVehicle Details:")
show_vehicle_details(car)  # Passing a standard Vehicle object

print("\nElectric Vehicle Details:")
show_vehicle_details(tesla)  # Passing an ElectricVehicle object
```

```python
# Exercise 4: Encapsulation

# Define the Vehicle class again, this time with encapsulation
class Vehicle:
    def __init__(self, make, model, year, mileage):
        self.make = make
        self.model = model
        self.year = year
        # Initialize the attributes, making mileage PRIVATE
        self.__mileage = mileage
    
    # Define get_mileage() method to access the private mileage attribute
    def get_mileage(self):
        return self.__mileage
    
    # Define set_mileage() method to update the mileage attribute
    def set_mileage(self, mileage):
        # Ensure the mileage cannot be set to a negative value
        if mileage >= 0:
            self.__mileage = mileage
        else:
            print("Mileage cannot be negative.")
            
    # Define display_info() method to display vehicle information
    def display_info(self):
        print(f"Make: {self.make}")
        print(f"Model: {self.model}")
        print(f"Year: {self.year}")
        print(f"Mileage: {self.get_mileage()} miles")

# Create an instance of Vehicle with encapsulated mileage
car_encapsulated = Vehicle("Honda", "Civic", 2020, 15000)

# Display the vehicle's initial information
car_encapsulated.display_info()

# Update mileage using the setter method
car_encapsulated.set_mileage(16000)

# Display the updated vehicle information
car_encapsulated.display_info()

# Attempt to set a negative mileage
car_encapsulated.set_mileage(-500)

# Display the vehicle information after the invalid update attempt
car_encapsulated.display_info()
```
